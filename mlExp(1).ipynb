```python
import gradio as gr
from model import load_model, predict

# 加载自主训练的模型
model = load_model("plant_disease_cnn.pth") 

def gradio_predict(image):
    # 自主实现预处理和推理
    processed_img = preprocess(image)
    pred_class, confidence, heatmap = predict(model, processed_img)
    
    # 生成可视化结果
    fig = plot_heatmap(image, heatmap)
    report = f"诊断结果：{pred_class}（置信度：{confidence:.2%}）"
    
    return report, fig

# 界面布局
with gr.Blocks(title="植物诊疗系统") as demo:
    gr.Markdown("# 🌱 植物病虫害检测系统")
    with gr.Row():
        input_img = gr.Image(label="上传叶片照片", type="pil")
        output_text = gr.Textbox(label="诊断报告")
        output_plot = gr.Plot(label="病变区域可视化")
    submit_btn = gr.Button("开始诊断")
    
    submit_btn.click(
        fn=gradio_predict,
        inputs=input_img,
        outputs=[output_text, output_plot]
    )
```

### **三、关键模块实现细节**
#### **1. 模型服务化改造**
```python
# model.py
import torch
from gradcam import GradCAM

def load_model(path):
    model = torch.load(path)
    model.eval()
    return model

def preprocess(image):
    transform = transforms.Compose([
        transforms.Resize(256),
        transforms.CenterCrop(224),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])
    return transform(image).unsqueeze(0)

def predict(model, input_tensor):
    with torch.no_grad():
        output = model(input_tensor)
    
    # 获取预测结果
    prob = torch.nn.functional.softmax(output, dim=1)
    confidence, pred_idx = torch.max(prob, 1)
    
    # 生成Grad-CAM
    target_layer = model.features[-2]  # 自定义模型的最后第二层
    cam = GradCAM(model, target_layer)
    heatmap = cam(input_tensor)
    
    return class_names[pred_idx], confidence.item(), heatmap
```

#### **2. Grad-CAM可视化实现**
```python
# gradcam.py
import matplotlib.pyplot as plt

def plot_heatmap(original_img, heatmap):
    fig, ax = plt.subplots()
    ax.imshow(original_img)
    ax.imshow(heatmap, alpha=0.5, cmap='jet')
    ax.axis('off')
    return fig
```
